##Question 2:

Let’s compare the EC values between Condition and Location. We’re going to look at three different statistical comparisons.

    Is there a statistically significant difference in EC between locations?

    Is there a statistically significant difference between in EC between saturation conditions?

    Is there a statistically significant difference between the interaction between saturation conditions and location and EC?


```r
#Reclassify both of our datasets to be factors
soils_df$Location<-factor(soils_df$Location)
soils_df$Condition<-factor(soils_df$Condition)

EC_Locations<-aov(EC_Av~Location, data=soils_df)
summary(EC_Locations)

EC_Conditions<-aov(EC_Av~Condition, data=soils_df)
summary(EC_Conditions)

EC_Int<-aov(EC_Av~Location*Condition,data=soils_df)
summary(EC_Int)
```
